** How many Job Titles were represented by only one person in 2013? (e.g. Job Titles with only one occurence in 2013?) **

** How many people have the word Chief in their job title? (This is pretty tricky) **

```python
df['JobTitle'].str.contains('Chief').sum() #it returns the all cheif contains count
```

```python
df['JobTitle'].str.contains('Chief',case = False).sum() #if the chief CHIEF like the actual content checked by it Uppercese lowercase all things
```

```python
df['JobTitle'].str.contains('Chief',case = False,na = False).sum()
```

```python
df['JobTitle'].str.contains(r'\bChief\b',case = False,na = False).sum()
```

```python
df['JobTitle'].str.split().apply(lambda x: 'Chief' in x).sum()
```

```python
df['JobTitle'].str.contains('chief',case = False,na = False).sum()
```

```python
df['JobTitle'].str.lower().str.split().apply(
    lambda x: 'chief' in x
).sum()
```
