```python
#contest 2 1
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'get_discount' function below.
#
# The function is expected to return a FLOAT.
# The function accepts INTEGER price as parameter.
#

def get_discount(price):
    # Write your code here
    
    if price < 1000:
        return price
    elif 1000 <= price < 5000:
        return price - ((price * 5)/100)
    elif 5000 <= price < 10000:
        return price - ((price * 10)/100)
    else:
        return price - ((price * 20)/100)

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    final_price = get_discount(n)

    fptr.write(str(final_price) + '\n')

    fptr.close()
```

```python
#contest 2 2
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'invert' function below.
#
# The function is expected to return a STRING.
# The function accepts STRING s1 as parameter.
#

def invert(s1):
    result = ""

    for i in s1:
        if i.isupper():
            result += i.lower()

        elif i.islower():
            result += i.upper()

        else:
            result += "*"

    return result
  

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = input()

    s2 = invert(n)

    fptr.write(s2 + '\n')

    fptr.close()
```

```python
x = 0
y = 0

while True:
    
    command = input()
    
    if command == 'STOP':
        break
    direction,steps = command.split()
    
    step = int(steps)
    
    if direction == 'UP':
        y += step
    elif direction == 'DOWN':
        y -= step
    elif direction == 'RIGHT':
        x += step
    elif direction == 'LEFT':
        x -= step
        
print(x,y)
```
