## Series Funtions

## 1) .unique()

```python
df['Address'].unique() # Gives the unique value from data
```

## 2) .nunique()

```python
df['Address'].nunique() #total of unique value
```

## 3) .value_counts()

```python
df['Address'].value_counts() #tells the count of unique values elements
```

## 4) .appy()

```python
df['Address'].apply(lambda x : x[:2]) #like we use map in list we use appy in series
```

```python
df['Address'].apply(lambda x : x[0] + x[-1])
```

```python
np.where(df['Marks'] >= 50, df['Marks'] + 5, df['Marks'] + 10)
```

```python
df['Marks'].apply(lambda x : x+ 10 if x <= 50 else x + 5)
```
