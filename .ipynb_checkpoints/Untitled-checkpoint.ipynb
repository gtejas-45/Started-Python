```python
import pandas as pd
```

```python
d1 = {'e_id':[101,102,103,104,105],
      'e_name':['amol','amit','akash','avinash','atish'],
      'e_sal':[25000,17500.5,55000,82000,42200.1],
      'e_dept':['HR','HR','IT','IT','IT']}
df = pd.DataFrame(d1,index=['A','B','C','X','Y'])
df
```

# iloc() is gives actual content of that index 

```python
df.iloc[3,2] #gives the value at that index
```

```python
#and we can update values also
df.iloc[3,2] = 6700
df
```

```python
df['e_sal'].mean()
```

```python
df.iloc[2:,:2] #rows and column
```

```python
print(df)
```

# .loc() gives the row series

```python
df.loc['A']
```

```python
df
```

```python
df['e_dept'] == 'IT'
```

```python
#data of employee of IT department
df[df['e_dept'] == 'IT']
```

```python
#we need only salary
df[df['e_dept'] == 'IT']['e_sal']
```

```python
#mean of salary
df[df['e_dept'] == 'IT']['e_sal'].mean()
```
