```python
np.random.seed(49)
marks = np.random.randint(0,100,(6,4))
```

```python
marks
```

```python
#topper marks in every subject?
print(marks.max(axis = 0))
```

```python
#what is avg marks of topper?
print(marks.max(axis = 0).mean())
```

```python
#addition of each row
print(marks.sum(axis = 1))
```

```python
#avg marks of each subject
print(marks.mean(axis = 0))
```
