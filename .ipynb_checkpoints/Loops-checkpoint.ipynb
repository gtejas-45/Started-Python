```python
#find degigit count 
s1 = 'A12 B3.14'
i = 0
count = 0
while i < len(s1):
    if s1[i].isdigit():
        count = count +1
    i = i +1
print(count)
    
```

```python
#find degigit sum
s1 = 'A12 B3.14'
i = 0
sum = 0
while i < len(s1):
    if s1[i].isdigit():
        sum = sum + int(s1[i])
    i = i +1
print(sum)
```

```python
cnt = 0
total = 0
for n in range(100,1000):
    copy = n
    t = 0
    while n != 0:
        d = n % 10
        n = n // 10
        t = t + (d**3)
    if t == copy:
        print("armstrong number is ",copy)
        cnt =cnt + 1
        total = total + copy
print(cnt)
print("sum",total)
        
    
```

```python
count = 0
total = 0
for n in range (1000,2000):
    copy = n
    rev = 0
   
    while n != 0:
        d = n % 10
        n = n // 10
        rev = d + (rev * 10)
    if rev == copy:
        print("Number is palindrome",copy)
        count = count + 1
        total = total + copy
print("total",count)
print("sum",total)
    
```

```python
n = 5
cnt = 0
for i in range(2,n):
    if n % i == 0:
        cnt = cnt + 1
if cnt == 0:
    print ('Prime Number')
else:
    print('Not Prime number')
```

```python
total = 0
for i in range(500,600):
    cnt = 0
    for j in range(2,i):
        if i % j == 0:
            cnt = cnt + 1
    if cnt == 0:
        print ('Prime Number ',i)
        total += 1
print(total)

    
```

```python
n = 10
cnt = 0
for i in range(2,n):
    if n % i == 0:
        print('multiplied by',i)
        cnt = cnt + 1
if cnt == 0:
    print ('Prime Number')
else:
    print('Not Prime number')
```

```python
#break & continue
i = 0
while i < 10:
    i = i + 1
    if i % 3 == 0:
        continue
    if i % 7 == 0:
        break
    print(i)
print('bye')
```

```python
n = 30
cnt = 0
for i in range(2,n//2+1):#we can use n//2 half the iteration
    if n % i == 0:
        print('multiplied by',i)
        cnt = cnt + 1
if cnt == 0:
    print ('Prime Number')
else:
    print('Not Prime number')
```

```python
n = 30
cnt = 0
for i in range(2,int(n**0.5) + 1):
    if n % i == 0:
        print('multiplied by',i)
        cnt = cnt + 1
if cnt == 0:
    print ('Prime Number')
else:
    print('Not Prime number')
```
