```python
name = ['Mr.Pratik Patil','Ms.Mayuri Rajmane','Mr.Gajanan Kulkarni','Mrs.Priya Rai']
#longest length name
m = len(name[0])
n = ""
for i in name:
    if len(i) > m:
       # m = len(i)
        n = i
        
print(n)
```

```python
#write a code to print surname only
for i in name:
    print(i.split(" ")[1])
```

```python
#write a code to print female names only
for i in name:
    if i.split()[0].startswith('Ms') or i.split()[0].startswith('Mrs'):
        print(i)
```
