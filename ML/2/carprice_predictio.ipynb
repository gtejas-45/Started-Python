```python
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
```

```python
print(df.shape)

df.info()

df.describe()
```

```python
df.columns = df.columns.str.strip()

df = df.drop(
    columns=["Unnamed: 0"],
    errors="ignore"
)

df = df.rename(columns={
    "ownsership": "ownership"
})

print(df.columns.tolist())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()

print("Shape after duplicate removal:", df.shape)
```

```python
df = df.drop_duplicates()

print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
df = df[
    (df["price(in lakhs)"] > 0) &
    (df["price(in lakhs)"] <= 200)
].copy()

print(df.shape)
```

```python
print(df["mileage(kmpl)"].max())
print(df["engine(cc)"].max())
print(df["max_power(bhp)"].max())
print(df["torque(Nm)"].max())
```

```python
df.loc[
    ~df["mileage(kmpl)"].between(1, 100),
    "mileage(kmpl)"
] = np.nan

df.loc[
    ~df["engine(cc)"].between(100, 8000),
    "engine(cc)"
] = np.nan

df.loc[
    ~df["torque(Nm)"].between(1, 3000),
    "torque(Nm)"
] = np.nan

df.loc[
    ~df["kms_driven"].between(0, 300000),
    "kms_driven"
] = np.nan
```

```python
df = df.drop(columns=["max_power(bhp)"])
```

```python
text_columns = [
    "car_name",
    "insurance_validity",
    "fuel_type",
    "ownership",
    "transmission"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print(df["fuel_type"].value_counts())
print(df["transmission"].value_counts())
print(df["ownership"].value_counts())
```

```python
numeric_columns = [
    "seats",
    "kms_driven",
    "manufacturing_year",
    "mileage(kmpl)",
    "engine(cc)",
    "torque(Nm)"
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
for col in text_columns:
    df[col] = df[col].replace("nan", np.nan)
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
current_year = pd.Timestamp.today().year

df["how_old"] = (
    current_year - df["manufacturing_year"]
)

print(
    df[["manufacturing_year", "how_old"]].head()
)
```

```python
x = df[
    [
        "car_name",
        "insurance_validity",
        "fuel_type",
        "seats",
        "kms_driven",
        "ownership",
        "transmission",
        "manufacturing_year",
        "how_old",
        "mileage(kmpl)",
        "engine(cc)",
        "torque(Nm)"
    ]
]

y = df["price(in lakhs)"]

print(x.head())
print(y.head())
```

```python
x = pd.get_dummies(x, dtype=int)

print(x.head())
print(x.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("Model training completed")
```

```python
y_pred = model.predict(X_test)

print("R2 Score:", r2_score(y_test, y_pred))

print(
    "MAE:",
    mean_absolute_error(y_test, y_pred)
)

print(
    "RMSE:",
    np.sqrt(mean_squared_error(y_test, y_pred))
)
```

```python
rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("Random Forest R2:", r2_score(y_test, rf_pred))

print(
    "Random Forest MAE:",
    mean_absolute_error(y_test, rf_pred)
)

print(
    "Random Forest RMSE:",
    np.sqrt(mean_squared_error(y_test, rf_pred))
)
```
