```python
import pandas as pd
df = pd.read_csv("../../datasets/Social_Network_Ads.csv")
df.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier()
```

```python
x = df[["Age","EstimatedSalary"]]
y = df["Purchased"]
```

```python
model.fit(x,y)
```

```python
model.score(x,y)
```

```python
model = KNeighborsClassifier(n_neighbors=3) #by default it is 5 we can give manually
model.fit(x,y)#It basically store the all co_ordinates in this of all colums
model.predict([[30,30000],[30,35000],[50,100000]])#it 1)find each point distance from the point 2)ascending sort 3)find numbers of closing point and give this result
```

```python
model.score(x,y)
```

```python
import seaborn as sns
sns.scatterplot(data = df,x="Age",y = "EstimatedSalary",hue = "Purchased")
```

# MinMaxScaling

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
scaler.fit(x)
x_scaled = scaler.transform(x)
```

```python
model.fit(x_scaled,y)
```

```python
model.score(x_scaled,y)
```

```python
#before Scaling the some point gives the wrong aswers
model.predict([[30,38000]])
```

```python
#by scaling we do correct it
q = [[30,30000],[30,35000],[50,100000]]
model.predict(q)
```

```python
q_scaled = scaler.transform(q)
model.predict(q_scaled)
```
