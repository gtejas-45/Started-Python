# IQR

```python
import numpy as np
marks = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(np.percentile(marks,25))
print(np.percentile(marks,50))
print(np.percentile(marks,75))
```

```python
q1 = np.percentile(marks,25)
q2 = np.percentile(marks,50)
q3 = np.percentile(marks,75)
```

```python
IQR = q3 - q1
print(IQR)
```

# Outliers _ the values far from the values 

```python
data = [10, 12, 13, 14, 15, 16, 18, 100]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = [
    x for x in data
    if x < lower_fence or x > upper_fence
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
print("Potential outliers:", outliers)
```

# by pandas

```python
import pandas as pd

df = pd.DataFrame({
    "Salary": [25000, 28000, 30000, 32000, 35000, 38000, 150000]
})

print("Q1:", df["Salary"].quantile(0.25))
print("Median:", df["Salary"].quantile(0.50))
print("Q3:", df["Salary"].quantile(0.75))
print("IQR:", df["Salary"].quantile(0.75) -
      df["Salary"].quantile(0.25))
```

```python
import matplotlib.pyplot as plt

data = [10, 12, 13, 14, 15, 16, 18, 100]

plt.boxplot(data)
plt.title("Box Plot to Identify Potential Outliers")
plt.ylabel("Values")
plt.show()
```
