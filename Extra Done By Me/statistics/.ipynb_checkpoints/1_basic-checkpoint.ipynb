```python
marks = [10,20,30,40,50,20,20,50,60]
```

# mean

```python
import numpy as np
print(np.mean(marks))
```

# median

```python
print(np.median(marks))
```

# mode

```python
from statistics import mode
print(mode(marks))
```

# range

```python
print(max(marks) - min(marks))
print(np.max(marks) - np.min(marks))
```

# by pandas

```python
import pandas as pd
s = pd.Series(marks)
print(s.mean())
print(s.median())
print(s.mode())
print(max(s) - min(s))
```

# using numpy

```python
print(np.var(marks)) #population varience
print(np.std(marks)) #Population standard deviation
```

```python
print(np.var(marks,ddof = 1)) #sample varience
print(np.std(marks, ddof = 1)) #sample standard deviation
```

# using pandas

```python
print(s.var())
print(s.std())
```

```python
import numpy as np

data = [2, 4, 6]

print("Sample variance:", np.var(data, ddof=1))
print("Sample standard deviation:", np.std(data, ddof=1))
```
