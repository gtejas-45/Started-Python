```python
a1 = '$50'     # '₹4750'
a2 = '€100'    # '₹11000'
a3 = '$2300'   #218500
a4 = '€15'     #1650
amount = a4

if amount.startswith('$'):
    C = amount.split('$')[1]
    amt = int(C)
    print('₹' +  f"{amt * 95}")
else:
    C = amount.split('€')[1]
    amt = int(C)
    print('₹', amt * 110)
    
```
