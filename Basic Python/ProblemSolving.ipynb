```python
address = ['Katraj,Pune','Dadar,Mumbai','KP,Delhi','Hinjewadi,Pune']
```

```python
#1)cities from above address
city = []
for i in address:
    city.append(i.split(',')[1])
city
```

```python
#2)create a list of adrees that belongs to the Pune
c = []
for i in address:
    if i.split(',')[1] == 'Pune':
        c.append(i)
c
```

```python
#3)create a list of states that belongs to maharastra
cities = ['Mumbai','Pune']
mh = []
for i in address:
    if i.split(',')[1] in cities:
        mh.append(i)
mh
        
```
