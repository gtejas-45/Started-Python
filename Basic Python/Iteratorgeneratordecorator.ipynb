```python
numbers = [10, 20, 30]

iterator = iter(numbers)

while True:
    try:
        x = next(iterator)
        print(x)
    except StopIteration:
        break
```

```python
class Count:

    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):

        if self.num <= 5:
            value = self.num
            self.num += 1
            return value

        raise StopIteration

obj = Count()
for x in obj:
    print(x)
```

# Generator : A generator is an easier way to create an iterator.

```python
#don't need to write __iter__ or __next__
def test():

    print("Start")

    yield 10

    print("Middle")

    yield 20

    print("End")

    yield 30
```

```python
test()
```

```python
g = test()
```

```python
next(g)
```

```python
next(g)
```

```python
next(g)
```

```python
next(g)
```

```python
#don't need to write __iter__ or __next__
def test():

    print("Start")

    yield 10

    print("Middle")

    yield 20

    print("End")

    yield 30
```

```python
g = test()
for i in g:
    print(i)#it don't give error
```

```python
def numbers():

    for i in range(1, 1000001):
        yield i
```

```python
g = numbers()
print(next(g))
```

```python
#generate only odd number
odd = (i for i in range(1,20) if i % 2 != 0)
for i in odd:
    print(i)
```
