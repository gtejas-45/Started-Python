```python
students = [{'name':'amit','marks':78,'fees':10000,'city':'pune'},
            {'name':'amol','marks':82,'fees':5000,'city':'mumbai'},
            {'name':'akash','marks':45,'city':'pune'},
            {'name':'anay','marks':52,'fees':20000,'city':'mumbai'}]
# 1. Find total marks of all students
# 2. Print name sof students from pune.
# 3. Find total fees collected.
```

```python
total = 0
for st in students:
    total = total + st['marks']
print(total)
```

```python
for st in students:
    if st['city'] == 'pune':
        print(st['name'])
```

```python
total = 0
for st in students:
    total = total + st.get('fees',0)
print(total)
```

```python
d1 = {}
for st in students:
    d1.setdefault(st['city'],[]).append(st['name'])
d1
```
