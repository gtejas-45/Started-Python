```python
def isArmstrong(n):
    m = n
    t = 0
    while n != 0:
        d = n % 10
        n = n // 10
        t = t + (d ** 3)
    if m == t:
        return True
    else:
        return False
```

```python
isArmstrong(153)
```

```python
cnt = 0
for i in range(300,400):
    if isArmstrong(i):
        cnt += 1
print(cnt)
```

```python
cnt = 0
for i in range(300,600):
    if isArmstrong(i):
        print(i)
        cnt += 1
print(cnt)
```

```python
def isPalindrome(n):
    copy = n
    rev = 0
    while n != 0:
        d = n % 10
        n = n // 10
        rev = d + (rev*10) 
    if rev == copy:
        return True
    else:
        return False
```

```python
isPalindrome(121)
```

```python
cnt = 0
for i in range(5000,6000):
    if isPalindrome(i):
        print(i)
        cnt += 1
print(cnt)
```

```python
sum = 0
for i in range(5000,6000):
    if isPalindrome(i):
        sum += i
print(sum)
```

```python
def isPrime(n):
    for i in range(2,int(n**0.5)+ 1):
        if n % i == 0:
            return False
            break
    return True
isPrime(6)
        
     
```

```python
#count of prime
cnt = 0
for i in range (10000,11000):
    if isPrime(i):
        cnt += 1
print(cnt)
```

```python
#first 10 prime number addition
cnt = 0
total = 0
for i in range (2,100):
    if isPrime(i):
        print(i)
        cnt += 1
        total += i
        if cnt == 10:
            break
print(total)
```

```python
def final_amt(amt, city = 'Pune'):
    if amt >10000:
        famt = amt - (amt * 0.1)
        if city == 'Pune':
            famt = famt - 500
        else:
            famt = famt - 200
            
    else:
        famt = amt - (amt * 0.05)
        if city == 'Pune':
            famt = famt - 500
        else:
            famt = famt - 200
    return famt
        
        
        
```
