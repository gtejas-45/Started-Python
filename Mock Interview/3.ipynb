```python
s = "I love python"

# 1. Flip the entire sentence backward
s1 = s[::-1]  # "nohtyp evol I"

# 2. Break it into a list of words
s2 = s1.split()  # ['nohtyp', 'evol', 'I']

# 3. Reverse the order of the list
s2.reverse()  # ['I', 'evol', 'nohtyp']

# 4. Glue the list back into a single sentence string
final_result = " ".join(s2)

print(final_result)
# Output: I evol nohtyp
```
