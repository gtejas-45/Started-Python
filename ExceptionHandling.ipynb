# **What is an Exception?**
##### An exception is an error that occurs while a program is running and interrupts the normal flow of the program.

```python
a = 10
b = 0

print(a / b)
```

```python
try:
    a = 10
    b = 0

    print(a / b)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

# **Common Python Exceptions**

```python
# 1. ZeroDivisionError -- Division by zero

try:
    a = 10
    b = 0

    print(a / b)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
# 2. ValueError
# Occurs when the type may be correct but the value is invalid.
try:
    age = int(input("Enter age: "))
    print(age)

except ValueError:
    print("Please enter a valid number")
```

```python
# 3. TypeError
# Occurs when an operation is performed on an inappropriate data type.
try:
    a = 10
    b = "20"
    
    print(a + b)
except TypeError:
    print("Addition is not possible")
```

```python
# 4. IndexError
try:
    numbers = [10, 20, 30]
    print(numbers[5])

except IndexError:
    print("Index does not exist")
```

```python
# 5. KeyError -- imp with dictinary
student = {
    "name": "Tejas",
    "age": 21
}
try:
    print(student["marks"])

except KeyError:
    print("Key does not exist")
```

# **Multiple Except Block**

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))

    print(a / b)

except ValueError:
    print("Please enter numbers only")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))

    print(a / b)

except ValueError:
    print("Please enter numbers only")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

# **else Block**
##### 1)Python also provides an else block.
##### 2)It runs when the try block runs successfully 
##### 3)Otherwise don't run -- runs the except block

```python
try:
    number = int(input("Enter number: "))

except ValueError:
    print("Invalid number")

else:
    print("You entered:", number)
```

```python
try:
    number = int(input("Enter number: "))

except ValueError:
    print("Invalid number")

else:
    print("You entered:", number)
```

```python
try:
    print("Hello")

except:
    print("Error")

else:
    print("Success") #else runs when try run
```

# **finally**

##### finally is a very important block.

###### It always executes, whether an exception occurs or not.

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program completed")
```

```python
try:
    print("Hello")

except:
    print("Error")

finally:
    print("Finally executed")
```

# **raise Keyword**
##### Sometimes we want to manually create an exception.

```python
age = 15

if age < 18:
    raise ValueError("Age must be 18 or above")
```

```python
try:
    age = 15

    if age < 18:
        raise ValueError("You are not eligible")

except ValueError as e:
    print(e)
```

# **Real Time Example - ATM**

```python
try:
    Balance = 50000
    Amount = int(input("Enter Amount to withdraw: "))

    if(Amount > Balance):
        raise ValueError("Insufficient balance")
    Balance = Balance - Amount
    print("Reminning balance",Balance)
    print("Withdraw Successful")
except Exception as e:
    print(e)
    
```

```python
try:
    Balance = 50000
    Amount = int(input("Enter Amount to withdraw: "))

    if(Amount > Balance):
        raise ValueError("Insufficient balance")
    Balance = Balance - Amount
    print("Reminning balance",Balance)
    print("Withdraw Successful")
except Exception as e:
    print(e)
```

# **Exception Handling With Functions**

```python
def divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        return "Cannot divide by zero"


print(divide(10, 2))
print(divide(10, 0))
```
