```python
### Students roll numbered from 1 to 20. phy, che, math
phy = {1,3,5,6,10,12,13,14,18}
che = {2,4,5,6,7,8,11,13,18,19}
math = {2,3,6,9,10,11,17,18,20}

# 1. Roll number opted for phy as well as maths => 3,6,10,18
# 2. Roll numbers opted for all 3 subjects => 6,18
# 3. Roll numbers opted for phy and che but not for math =>5,13,
# 4. Roll numbers opted for exactly 2 subjects =>5,13,3,10,2,11
# 5. Roll numbers not opted any subject => 15,16
# (Hint ALL = set(range(1,21)))
```

```python
phy.intersection(math)
```

```python
phy.intersection(math).intersection(che)
```

```python
phy.intersection(che).difference(math)
```

```python
s1 = phy.intersection(math)
s2 = math.intersection(che)
s3 = che.intersection(phy)
s4 = s1.union(s2).union(s3)
s5 = phy.intersection(math).intersection(che)
s4.difference(s5)
```
