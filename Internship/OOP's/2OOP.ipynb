```python
class Student:
    def __init__ (self, roll_num, name, marks):
        self.roll_num = roll_num
        self.name = name
        self.marks = marks
    def isTopper(self, marks):
        if(self.marks > 85):
            print("Topper")
        else:
            print("Not Topper")
    
Student_one1 = Student(1,"Raj",90)
print(Student_one1.roll_num)
print(Student_one1.name)
print(Student_one1.marks)
Student_one1.isTopper(Student_one1.marks)
```

```python
class Car:
    brand = "Toyota"
    color = "White"
    speed = 120

    def drive(self):
        print("Car is driving.")
    
car1 = Car()
print(car1.brand)
print(car1.color)
print(car1.speed)
car1.drive()
```

```python
class Teacher:
    def __init__ (self, id, name, Experience):
        self.id = id
        self.name = name
        self.Experience = Experience
    def isHeBecomePrinciple(self, Experience):
        if(self.Experience > 19):
            print("Yes")
        else:
            print("No")
    
Teacher_1 = Teacher(1,"Prof.S.K.Joshi",25)
print(Teacher_1.id)
print(Teacher_1.name)
print(Teacher_1.Experience)
Teacher_1.isHeBecomePrinciple(Teacher_1.Experience)
```

```python
class BankAccount:
    account_number = 12345
    holder_name = "Tejas"
    balance = 50000

    def deposit(self,amt):
        self.balance = self.balance + amt
        print("Money deposited successfully and Balance now is ", self.balance)
        
emp = BankAccount()
print(emp.account_number)
print(emp.balance)
emp.deposit(1200)
```
