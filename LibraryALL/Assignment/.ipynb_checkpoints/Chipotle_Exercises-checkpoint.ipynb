### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv).

```python
df = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv',sep='\t')
```

### Step 3. Assign it to a variable called chipo.

```python
df.head(30)
```

### Step 4. How many products cost more than $10.00?

```python
df['item_price_float'] = df['item_price'].apply(lambda price:float(price[1:]))
```

```python
# df['item_price_float'] = df['item_price_float']/df['quantity']
```

```python
df[df['item_price_float']>10]['item_name'].nunique()
```

### Step 5. What is the price of each item?
###### print a data frame with only two columns item_name and item_price

```python
df[["item_name","item_price"]]
```

### Step 6. Sort by the name of the item

```python
df.sort_values(by = "item_name")
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
sum(df[df["item_price_float"].max() == df["item_price_float"]]["quantity"])
```

```python
price = df["item_price"].str.replace("$", "", regex=False).astype(float)

df.loc[price == price.max(), "quantity"].sum()
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
sum(df["item_name"] == "Veggie Salad Bowl")
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
sum((df["item_name"] == "Canned Soda") & (df["quantity"] > 1))
```

### Step 10. What is the order id of order with the highest total price of all items ordered in that order?

```python
# df[df["item_price_float"].max() == df["item_price_float"]]["order_id"]
```

```python
df.groupby("order_id")["item_price_float"].sum().idxmax()
```

### Step 11. In How many orders, atleast one items is ordered multiple times?

```python
df[df["quantity"] > 1]["order_id"].nunique()
```

### Step 12. Create a billing report (dataframe) which has columns as 'Order_Id', 'Total_Bill', 'Total Bill After Discount'. Restaurant owner has decided to give discount of 5% on orders of above $50 total cost.

```python
df_billingreport = pd.DataFrame(
    {
        "Total_Bill" : df["order_id"],
        "Total_Bill" : df["item_price_float"]
        
    }
)
```

```python
df_billingreport
```

```python
df_billingreport["Total Bill After Discount"] = df_billingreport["Total_Bill"].apply(lambda x : (x * 0.95) if x > 50 else x)
```

```python
df_billingreport
```

```python
sum(df_billingreport["Total_Bill"] > 50)
```
