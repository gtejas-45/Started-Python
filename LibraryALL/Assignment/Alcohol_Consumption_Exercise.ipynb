### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv). 

```python
df = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv")
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.info()
```

### Step 3. Assign it to a variable called drinks.

```python
drinks = df
drinks.head()
```

### Step 4. Which continent drinks more beer on average?

```python
drinks.groupby("continent")["beer_servings"].mean().sort_values(ascending=False)
```

```python
max_beer = drinks.groupby("continent")["beer_servings"].mean().sort_values(ascending=False)
print(max_beer.idxmax(),max_beer.max())
```

### Step 5. For each continent print the statistics for wine consumption.

```python
drinks.groupby("continent")["wine_servings"].describe()
```

```python
drinks.groupby("continent")["wine_servings"].agg(["mean","median","std","max","min"])
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
df.groupby("continent")[["total_litres_of_pure_alcohol","beer_servings","spirit_servings","wine_servings","total_litres_of_pure_alcohol"]].mean()
```

```python
drinks.groupby("continent").mean(numeric_only=True)
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
drinks.groupby("continent").median(numeric_only=True)
```

### Step 8. Find out (list) the names of countries which shows no cosumption of any kind of drink

```python
df.head()
```

```python
zero_consumption = drinks[
    (drinks["beer_servings"] == 0) &
    (drinks["spirit_servings"] == 0) &
    (drinks["wine_servings"] == 0) &
    (drinks["total_litres_of_pure_alcohol"] == 0)
]

print(zero_consumption["country"].to_list())
```

### Step 9. Find out (list) the names of countries which has more wine_servings than beer_servings from Europe

```python
sum(df["continent"].str.lower() == "eu")
```

```python
df[(df["wine_servings"] > df["beer_servings"]) & (df["continent"] == "EU")]["country"]
```

```python
df[(df["wine_servings"] > df["beer_servings"]) & (df["continent"] == "EU")]["country"].to_list()
```

### Step 10. Find out (list) the names of countries from Asia which has more beer consumption than avg beer_consumption of Europe

```python
df.groupby("continent")["beer_servings"].mean()
```

```python
s7 = df.groupby("continent")["beer_servings"].mean().iloc[2]
```

```python
df[(df["continent"] == "AS") & (df["beer_servings"] > s7)]["country"].to_list()
```
