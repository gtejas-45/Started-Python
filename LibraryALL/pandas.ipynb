### Merge can be done with only two dataset at a time but conacate done with many dataset

### Concate done with many datasets

```python
df1 = pd.DataFrame({
    'Name': ['Amit', 'Rahul'],
    'Age': [20, 21]
})

df2 = pd.DataFrame({
    'Name': ['Sneha', 'Priya'],
    'Age': [22, 23]
})

result = pd.concat([df1, df2])

print(result)
```

```python
result = pd.concat([df1, df2], ignore_index=True)
print(result)
```

```python
pd.concat([df1, df2], axis=0)
```

```python
pd.concat([df1, df2], axis=1)
```

# join() in Pandas

# join() is another way of combining DataFrames.

# It is mainly useful when you want to combine using the index.

```python
df1 = pd.DataFrame({
    'Name': ['Amit', 'Rahul']
}, index=[1, 2])

df2 = pd.DataFrame({
    'Marks': [80, 75]
}, index=[1, 2])
```

```python
result = df1.join(df2)

print(result)
```
