## Pivot_table

```python
# import pandas as pd

sales_data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop",
                "Phone", "Tablet", "Laptop", "Phone"],
    "City": ["Pune", "Mumbai", "Pune", "Delhi",
             "Mumbai", "Delhi", "Pune", "Pune"],
    "Sales": [65000, 25000, 18000, 72000,
              30000, 22000, 58000, 27000],
    "Quantity": [2, 5, 3, 1, 4, 2, 3, 6]
}

sales = pd.DataFrame(sales_data)

print(sales)
```

```python
#sum
pivot = pd.pivot_table(
    sales,
    index = "City",
    columns = "Product",
    values = "Sales",
    aggfunc = "sum",
    fill_value = 0
)
pivot   
```

```python
#avg
pivot1 = pd.pivot_table(
    sales,
    index = "City",
    columns = "Product",
    values = "Sales",
    aggfunc = "mean",
    fill_value = 0
)
pivot1 
```

```python
#count
pivot = pd.pivot_table(
    sales,
    index = "City",
    columns = "Product",
    values = "Sales",
    aggfunc = "count",
    fill_value = 0
)
pivot 
```

```python
# Managers often want to see both individual totals and the overall total.
pivot = pd.pivot_table(
    sales,
    index="City",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name = "Total"
)

print(pivot)
```

```python
pivot = pd.pivot_table(
    sales,
    index="City",
    columns="Product",
    values="Sales",
    aggfunc=["sum", "mean"]
)

print(pivot)
```

```python
result = pd.crosstab(sales["City"], sales["Product"])

print(result)
```

```python
result = pd.crosstab(sales["City"], sales["Product"],margins = True)

print(result)
```

```python
result = pd.crosstab(
    sales["City"],
    sales["Product"],
    normalize="index"
)

print(result.round(2))
```

```python
# 1. Create a pivot table showing total sales by city.
pd.pivot_table(sales, index="City", values="Sales", aggfunc="sum")
```

```python
# 2. Show total sales by city and product.
pd.pivot_table(sales, index="City", columns = "Product",values="Sales", aggfunc="sum")
```

```python
# 3. Find the average sales by product.
pd.pivot_table(sales, index="Product", values="Sales", aggfunc="mean")
```

```python
# 4. Create a pivot table with grand totals.
pivot = pd.pivot_table(sales, index="City", columns="Product", values="Sales", aggfunc="sum", fill_value=0, margins=True,margins_name="grand_total")
pivot
```

```python
# 5. Count records for each city and product.
pd.crosstab(sales["City"],sales["Product"])
```

```python
# 6. Show row-wise percentages for each city and product.
pd.crosstab(sales["City"], sales["Product"], normalize="index").round(2)
```

```python
# 7. Calculate total quantity by city and product.
pd.crosstab(sales["City"], sales["Product"], margins=True)
```

```python
import pandas as pd

sales = pd.Series([10, 20, 30, 40, 50, 60, 70, 80])

sales_group = pd.qcut(
    sales,
    q=4,#equal divide
    labels=["Q1", "Q2", "Q3", "Q4"]
)

result = pd.DataFrame({
    "Sales": sales,
    "Sales_Group": sales_group
})

print(result)
```
