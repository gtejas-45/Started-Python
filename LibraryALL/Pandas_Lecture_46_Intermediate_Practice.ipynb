**Tip:** For multiple conditions, use `&` for AND and `|` for OR. Put each condition in parentheses.

```python
import pandas as pd

data = {
    "Name": ["Tejas", "Rahul", "Amit", "Priya", "Neha", "Karan", "Sneha", "Om"],
    "Age": [21, 22, 20, 21, 23, 20, 22, 24],
    "Marks": [85, 65, 92, 45, 78, 35, 88, 72],
    "City": ["Pune", "Mumbai", "Nashik", "Pune", "Mumbai", "Nashik", "Pune", "Aurangabad"],
    "Department": ["Computer", "IT", "Computer", "ENTC", "IT", "Computer", "ENTC", "IT"],
    "Attendance": [92, 80, 96, 68, 85, 55, 91, 75]
}

df = pd.DataFrame(data)
df
```

## Q1 — Filter by marks

Display only students who scored more than 75 marks.

```python
# Write your solution below
df[df['Marks'] > 75]
```

## Q2 — Filter by age

Display students whose age is 21 or above.

```python
# Write your solution below
df[df['Age'] >= 21]
```

## Q3 — Filter by city

Display only students from Pune.

```python
# Write your solution below
df[df["City"]=="Pune"]
```

## Q4 — OR condition

Display students from Pune OR Mumbai.

```python
# Write your solution below
df[(df["City"] == "Pune") | (df["City"] == "Mumbai")]
```

## Q5 — AND condition

Display students who scored more than 70 AND have attendance of at least 80.

```python
# Write your solution below
df[(df["Marks"] > 70) & (df["Attendance"] >= 80)]
```

## Q6 — Multiple conditions

Display students who are in the Computer department AND scored at least 80 marks.

```python
# Write your solution below
df[(df["Department"] == "Computer") & (df["Marks"] >= 80)]
```

## Q7 — Select chosen columns

Display only Name, Department, and Marks for students who scored at least 75.

```python
# Write your solution below
df[df["Marks"] >= 75][["Name","Department","Marks"]]
```

## Q8 — Sort descending

Sort the complete DataFrame by Marks from highest to lowest.

```python
# Write your solution below
# df[df["Marks"].sort_values(ascending=False)]
df.sort_values(by='Marks', ascending=False)
```

## Q9 — Sort by two columns

Sort by Department first, then by Marks from highest to lowest within each department.

```python
# Write your solution below
df.sort_values(by = "Department").sort_values(by = "Marks",ascending=False)
```

## Q10 — Add a calculated column

Create a new column BonusMarks by adding 5 to each student's Marks. Keep the original Marks column.

```python
# Write your solution below
df["BonusMarks"] = df["Marks"] + 5
df
```

## Q11 — Add a result column

Create a column Result containing 'Pass' if Marks are at least 40, otherwise 'Fail'. Use a beginner-friendly method you know.

```python
# Write your solution below
df["Result"] = df["Marks"].apply(lambda x : "Pass" if x >= 40 else "Fail")
df
```

## Q12 — Rename a column

Rename Attendance to AttendancePercent and display the column names.

```python
# Write your solution below
df.rename(columns = {"Attendance": "AttendancePercent"},inplace=True)
df.columns
```

## Q13 — Remove a column

Create a DataFrame without the Age column. Do not modify the original df.

```python
# Write your solution below
df_1 = df[['Name','Marks','City','Department']]
df_1
```

## Q14 — Range filter

Display students whose Marks are between 60 and 90, including 60 and 90.

```python
# Write your solution below
df[df["Marks"].between(60,90)]
```

## Q15 — Membership filter

Display students whose Department is either IT or ENTC.

```python
# Write your solution below
df[df["Department"].isin(["IT","ENTC"])]
```

## Q16 — Combined challenge

Display Name, City, Marks, and Attendance only for students from Pune or Mumbai who scored at least 75 marks AND have attendance of at least 80.

```python
# Write your solution below
a = df["City"].isin(["Pune","Mumbai"]) 
b = df["Marks"] >= 75
c = df["AttendancePercent"] >= 80
df[a & b & c]
```
