```python
import numpy as np
```

```python
names = np.array(["A", "B", "C", "D"])
marks = np.array([75, 95, 60, 85])
# Top student's name
# Top student's marks
# Lowest student's name
# Students sorted by marks
# Students who scored above 80
```

```python
top = np.argmax(marks)#it returns the index of max elemnt
print(top)
```

```python
print(names[top])
print(marks[top])
```

```python
low = np.argmin(marks)#it returns the index of max elemnt
print(low)
```

```python
print(names[low])
print(marks[low])
```

```python
np.sort(marks)
```

```python
np.argsort(marks) #sort by index
```

```python
names[np.argsort(marks)]
```

```python
arr = np.array([10, 20, 30, 40, 50])
# Is any value greater than 45?
# Are all values greater than 5?
# Are all values greater than 20?
```

```python
data = np.array([10, 20, 30, 40, 50, 60, 70])

# 1. Find square root
# 2. Find square of every value
# 3. Find absolute values of [-10, 20, -30]
# 4. Round [2.345, 5.678] to 2 decimals
# 5. Find floor and ceil of [2.3, 5.8]
# 6. Find mean
# 7. Find median
# 8. Find variance
# 9. Find standard deviation
# 10. Find 25th and 75th percentile
```

```python
print(np.sqrt(data))
print(np.pow(data,2))
print(np.abs([-10, 20, -30]))
print(np.round( [2.345, 5.678],2))
print(np.floor([2.3, 5.8]))
print(np.ceil([2.3, 5.8]))
print(np.mean(data))
print(np.var(data))
print(np.std(data))
print(np.percentile(data,25))
print(np.percentile(data,75))
print(np.percentile(data,50))
```

```python

A = np.array([
    [2, 3],
    [4, 5]
])

B = np.array([
    [1, 2],
    [3, 4]
])

# 1. Add A and B
# 2. Subtract B from A
# 3. Multiply A and B element-wise
# 4. Perform matrix multiplication
# 5. Find transpose of A
# 6. Find determinant of A
# 7. Find inverse of A
# 8. Find dot product of [2,3] and [4,5]
```

```python
print(A + B)
print(A - B)
print(A * B)
print(A @ B)
print(A.T)
print(np.linalg.det(A))
print(np.linalg.inv(A))
print(np.dot([2,3],[4,5]))
```
