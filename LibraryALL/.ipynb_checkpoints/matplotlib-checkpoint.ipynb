```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [10000, 15000, 12000, 18000, 22000, 25000]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")

plt.grid(True)
plt.show()
```

```python
plt.plot(
    months,
    sales,
    color="green",
    marker="o",
    linestyle="--",
    linewidth=2,
    markersize=7
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.grid(True)
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptop_sales = [100, 120, 115, 150, 180]
phone_sales = [150, 140, 170, 190, 210]

plt.plot(months, laptop_sales, marker="o", label="Laptops")
plt.plot(months, phone_sales, marker="s", label="Phones")

plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Units Sold")

plt.legend()#it maily used to show the for which this line used with color
plt.grid(True)

plt.show()
```

```python

import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [50, 80, 30, 60]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    products,
    sales,
    color="skyblue",
    edgecolor="black"
)

plt.title("Product Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales Units")
plt.grid(axis="y", linestyle="--", alpha=0.5)#by y axis & linestlye(--)  & alpha is opacity

plt.bar_label(bars, padding=3) #adding labels to bar with padding 

plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

products = ["Laptop", "Mobile", "Tablet"]

pune_sales = [50, 80, 30]
mumbai_sales = [60, 70, 45]

x = np.arange(len(products))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(x - width/2, pune_sales, width, label="Pune")
plt.bar(x + width/2, mumbai_sales, width, label="Mumbai")

plt.xticks(x, products)
plt.title("Product Sales: Pune vs Mumbai")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "City": ["Pune", "Mumbai", "Nashik", "Nagpur", "Thane"],
    "Revenue": [250000, 400000, 180000, 300000, 350000]
})

# Sort revenue from highest to lowest
df = df.sort_values("Revenue", ascending=False)

print(df)

df.plot(
    x="City",
    y="Revenue",
    kind="bar",
    legend=False,
    title="Revenue by City",
    figsize=(8, 5)
)

plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Question 4 — Challenge
# How would you compare the sales of two companies across four cities using a grouped bar chart?
city = ["Pune","mumbai","Latur","omerga"]
comp1 = [25,36,18,29]
comp2 = [15,56,29,48]
plt.bar(city,comp1)
plt.plot(city,comp2)
plt.legend()
plt.show()
```

```python
plt.plot(city,comp1)
plt.plot(city,comp2)
plt.legend()
plt.show()
```

```python
# Create an array of positions for the bars
x = np.arange(len(city))  # [0, 1, 2, 3]
width = 0.35  # Width of each bar

# Plot bars side-by-side
plt.bar(x - width/2, comp1, width, label='Company 1')
plt.bar(x + width/2, comp2, width, label='Company 2')
plt.legend()
plt.show()
```

# Pie Charts and Percentage Analysis

```python
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [40, 30, 20, 10]

plt.pie(sales, labels=products)#labels by product

plt.title("Product Sales Distribution")
plt.show()
```

```python

import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [40, 30, 20, 10]

plt.figure(figsize=(7, 7))

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",#display the pecentage in block
    startangle=90,
    wedgeprops={"edgecolor": "white", "linewidth": 2}
)

plt.title("Product Sales Distribution")
plt.show()
```

```python
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [40, 30, 20, 10]

explode = (0.5, 0.2, 0, 0)#remove the high part from others by 0.5 and seccond high by 0.2

plt.pie(
    sales,
    labels=products, #label added 
    autopct="%1.1f%%", #give % 1.1 decimal if 1.2 last 2 digit
    pctdistance=0.7, #places percentages closer to the center.
    labeldistance=1.1, #places category labels slightly outside the pie.
    explode=explode,
    startangle=90 #
)

plt.title("Sales Distribution ")
plt.show()
```

```python

import matplotlib.pyplot as plt

expenses = [25000, 15000, 10000, 5000]
categories = ["Rent", "Salary", "Marketing", "Electricity"]

plt.figure(figsize=(7, 7))

plt.pie(
    expenses,
    autopct="%1.1f%%",
    startangle=90
)

plt.legend(
    categories,
    title="Expense Categories",
    loc="best"
)

plt.title("Company Expense Distribution")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Category": ["Rent", "Salaries", "Marketing", "Travel", "Other"],
    "Expense": [20000, 40000, 10000, 15000, 5000]
})

total = df["Expense"].sum()

df["Percentage"] = df["Expense"] / total * 100

print(df)
print("Total expense:", total)

plt.figure(figsize=(8, 8))

plt.pie(
    df["Expense"],
    labels=df["Category"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Monthly Company Expenses")
plt.show()
```
