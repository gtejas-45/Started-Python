```python

import matplotlib.pyplot as plt
import numpy as np

products = ["Laptop", "Mobile", "Tablet"]

pune_sales = [50, 80, 30]
mumbai_sales = [60, 70, 45]

x = np.arange(len(products))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(x - width/2, pune_sales, width, label="Pune")
plt.bar(x + width/2, mumbai_sales, width, label="Mumbai")

plt.xticks(x, products)
plt.title("Product Sales: Pune vs Mumbai")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

# Pie Charts and Percentage Analysis

```python
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [40, 30, 20, 10]

plt.pie(sales, labels=products)#labels by product

plt.title("Product Sales Distribution")
plt.show()
```

```python

import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [40, 30, 20, 10]

plt.figure(figsize=(7, 7))

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",#display the pecentage in block
    startangle=90,
    wedgeprops={"edgecolor": "white", "linewidth": 2}
)

plt.title("Product Sales Distribution")
plt.show()
```

```python
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [40, 30, 20, 10]

explode = (0.5, 0.2, 0, 0)#remove the high part from others by 0.5 and seccond high by 0.2

plt.pie(
    sales,
    labels=products, #label added 
    autopct="%1.1f%%", #give % 1.1 decimal if 1.2 last 2 digit
    pctdistance=0.7, #places percentages closer to the center.
    labeldistance=1.1, #places category labels slightly outside the pie.
    explode=explode,
    startangle=90 #
)

plt.title("Sales Distribution ")
plt.show()
```

```python

import matplotlib.pyplot as plt

expenses = [25000, 15000, 10000, 5000]
categories = ["Rent", "Salary", "Marketing", "Electricity"]

plt.figure(figsize=(7, 7))

plt.pie(
    expenses,
    autopct="%1.1f%%",
    startangle=90
)

plt.legend(
    categories,
    title="Expense Categories",
    loc="best"
)

plt.title("Company Expense Distribution")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Category": ["Rent", "Salaries", "Marketing", "Travel", "Other"],
    "Expense": [20000, 40000, 10000, 15000, 5000]
})

total = df["Expense"].sum()

df["Percentage"] = df["Expense"] / total * 100

print(df)
print("Total expense:", total)

plt.figure(figsize=(8, 8))

plt.pie(
    df["Expense"],
    labels=df["Category"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Monthly Company Expenses")
plt.show()
```

```python

import pandas as pd

data = pd.Series([10, 12, 13, 14, 15, 16, 18, 20, 22, 100])

q1 = data.quantile(0.25)
q3 = data.quantile(0.75)

iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

outliers = data[
    (data < lower_fence) |
    (data > upper_fence)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
print("Potential outliers:", outliers.tolist())
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]
expenses = [60, 90, 80, 100, 120]

fig, ax = plt.subplots(1, 2, figsize=(10, 4))#subplot(1,2) 1 row 2 column #fig is whole figure size

#main diff - plt.plot not we use ax[0], ax[1]
ax[0].plot(months, sales, marker="o")#ax[0] - 1st plot
ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

ax[1].bar(months, expenses) #second in same row a[1]
ax[1].set_title("Monthly Expenses")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")

plt.tight_layout() #tight_layout() helps prevent labels and titles from overlapping.
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 7))
ax[0, 0].plot(months, sales)
ax[0, 1].bar(months, expenses)
ax[1, 0].hist(sales)
ax[1, 1].scatter(sales, expenses)
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [1, 2, 2, 3, 4, 5, 6, 7],
    "Attendance": [45, 50, 55, 60, 70, 80, 85, 95],
    "Assignment_Score": [30, 40, 42, 50, 65, 72, 85, 92],
    "Marks": [25, 35, 38, 48, 60, 75, 82, 95]
}

df = pd.DataFrame(data)

print(df)
print(df.describe())
```

```python
# What patterns can we observe in student performance, study hours, and attendance?
plt.plot(df["Marks"],df["Study_Hours"],marker = "o")
plt.plot(df["Marks"],df["Attendance"],marker = "o")
plt.legend()
plt.show()
```

```python
# Line chart: Study hours vs. marks
plt.plot(df["Study_Hours"],df["Marks"],marker = "d")
# What to investigate: Do marks tend to increase as study hours increase?
```

```python
# Bar chart: Marks by student
plt.bar(df["Student"],df["Marks"])
plt.title("Marks by Student")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
# What to investigate: Which student scored the highest marks?
```

```python
# Histogram: Distribution of marks
plt.figure(figsize=(7, 4))

plt.hist(df["Marks"], bins=5, edgecolor="black")

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
# What to investigate: How are the marks distributed across the students?
```

```python
# Scatter plot: Attendance vs. marks
plt.scatter(df["Attendance"],df["Marks"],marker = "s",color = "pink")
plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
# What to investigate: Do students with higher attendance tend to have higher marks in this dataset?
```

```python
# Heatmap: Correlation matrix
corr = df.corr(numeric_only=True)
print(corr)
```

```python
sns.heatmap(
    corr  
)
```

```python
corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Student Performance Correlation")
plt.show()
```

```python
# Display four charts in one figure
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(df["Study_Hours"], df["Marks"], marker="o")
ax[0, 0].set_title("Study Hours vs Marks")

ax[0, 1].bar(df["Student"], df["Marks"])
ax[0, 1].set_title("Marks by Student")

ax[1, 0].hist(df["Marks"], bins=5, edgecolor="black")
ax[1, 0].set_title("Marks Distribution")

ax[1, 1].scatter(df["Attendance"], df["Marks"])
ax[1, 1].set_title("Attendance vs Marks")

plt.tight_layout()
plt.show()
```
