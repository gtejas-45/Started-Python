```python
import pandas as pd
sales_data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Tablet", "Laptop", "Phone"],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Delhi", "Pune", "Pune"],
    "Sales": [65000, 25000, 18000, 72000, 30000, 22000, 58000, 27000],
    "Quantity": [2, 5, 3, 1, 4, 2, 3, 6]
}

sales = pd.DataFrame(sales_data)
sales
```

```python
# Question 1: Total sales by city
s = sales.groupby("City")["Sales"].sum()
s
```

```python
print(s[s.max() == s])
```

```python
# Question 2: Total quantity sold for each product
sales.groupby("Product")["Quantity"].sum()
```

```python
sales.groupby("Product")["Quantity"].sum().idxmax()
```

```python
# Question 3: Average sales by product
sales.groupby("Product")["Sales"].mean()
```

```python
# Question 4: Total sales for each product in each city
sales.groupby(["Product","City"])["Sales"].sum()
```

```python
# How do you calculate total sales by city, from highest to lowest?
sales.groupby("City")["Sales"].sum().sort_values(ascending = False)
```

```python
# Q7. How do you calculate multiple statistics together?
sales.groupby("City")["Sales"].agg(["mean","median","max"])
```

```python
# import pandas as pd

data = {
    "Name": ["Tejas", "Rahul", "Priya", "Amit", "Neha", "Karan", "Sneha", "Om"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales", "IT", "HR"],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune", "Nashik", "Mumbai"],
    "Salary": [30000, 25000, 40000, 28000, 32000, 35000, 45000, 27000],
    "Age": [22, 24, 23, 25, 26, 24, 22, 25]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Calculate the average age for each department.
df.groupby("Department")["Age"].mean()
```

```python
# Calculate the total salary for each city.
df.groupby("City")["Salary"].sum()
```

```python
# Find the highest salary in each department.
df.groupby("Department")["Salary"].max()
```

```python
# Count employees in each city.
df.groupby("City")["Name"].count()
```

```python
df.groupby("City").size()
```

```python
# Calculate average salary and maximum salary by department.
df.groupby("Department")["Salary"].agg(["mean","max"])
```

```python
# Find total salary by department and city.
df.groupby(["Department","City"])["Salary"].sum().reset_index
```

```python
# Sort departments by total salary, highest first.
df.groupby("Department")["Salary"].sum().sort_values(ascending = False)
```

```python
# Display all employees in departments whose average salary is above ₹30,000.
df.groupby("Department").filter(lambda group: group["Salary"].mean() > 30000)
```

```python
customers = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4],
    "CustomerName": ["Tejas", "Rahul", "Priya", "Amit"],
    "City": ["Pune", "Mumbai", "Nashik", "Pune"]
})

orders = pd.DataFrame({
    "OrderID": [501, 502, 503, 504],
    "CustomerID": [1, 2, 1, 5],
    "Amount": [1500, 2500, 3000, 1200]
})
```

```python
# Task 1: Find orders with customer details
pd.merge(customers,orders,on ="CustomerID",how= "inner")
```

```python
# Task 2: Keep all customers, including those without orders
pd.merge(customers,orders,on ="CustomerID",how= "left")
```

```python
# Task 3: Calculate total order amount by customer
res = pd.merge(customers,orders,on ="CustomerID",how= "inner")
```

```python
res.groupby("CustomerID")["Amount"].sum()
```

```python
# How do you stack two DataFrames vertically?
pd.concat([customers, orders], axis=0, ignore_index=True)
```

```python
pd.concat([customers, orders], axis=1, ignore_index=True)
```
