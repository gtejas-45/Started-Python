## 1. Import NumPy

```python
import numpy as np

print(np.__version__)
```

## 2. Creating NumPy Arrays

### Main function
`np.array()` converts Python lists/tuples into NumPy arrays.

```python
arr = np.array([10, 20, 30, 40])
print(arr)
print(type(arr))
```

## 3. Array Dimensions and Properties

Important properties:
- `ndim` → number of dimensions
- `shape` → rows and columns / dimensions
- `size` → total number of elements
- `dtype` → data type

```python
arr1 = np.array([10, 20, 30])
arr2 = np.array([[10, 20, 30], [40, 50, 60]])

print("1D:", arr1)
print("ndim:", arr1.ndim)
print("shape:", arr1.shape)
print("size:", arr1.size)
print("dtype:", arr1.dtype)

print()
print("2D:", arr2)
print("ndim:", arr2.ndim)
print("shape:", arr2.shape)
print("size:", arr2.size)
print("dtype:", arr2.dtype)
```

## 4. 1D, 2D and 3D Arrays

```python
a1 = np.array([1, 2, 3])

a2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

a3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(a1)
print(a2)
print(a3)
print("Dimensions:", a1.ndim, a2.ndim, a3.ndim)
```

## 5. Indexing

### 1D
`arr[index]`

### 2D
`arr[row, column]`

Python indexing starts from `0`.

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[-1])

data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data[0, 1])
print(data[2, 2])
print(data[1])
```

## 6. Slicing

Pattern:

`arr[start:stop:step]`

- start included
- stop excluded
- step controls the jump

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print(arr[1:5])
print(arr[:4])
print(arr[2:])
print(arr[::2])
print(arr[::-1])

data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data[0:2, 1:3])
print(data[:, 1])
print(data[1, :])
```

## 7. Boolean Indexing / Filtering

```python
marks = np.array([45, 60, 75, 30, 90, 55])

print(marks[marks > 50])
print(marks[marks >= 60])
print(marks[(marks > 40) & (marks < 80)])
print(marks[(marks < 40) | (marks > 80)])
print(marks[~(marks > 50)])
```

## 8. View vs Copy

- `view()` shares the original data.
- `copy()` creates independent data.

```python
arr = np.array([10, 20, 30, 40])

view_arr = arr.view()
copy_arr = arr.copy()

view_arr[0] = 100
copy_arr[1] = 200

print("Original:", arr)
print("View:", view_arr)
print("Copy:", copy_arr)
```

## 9. Special Array Creation Functions

- `zeros()`
- `ones()`
- `full()`
- `arange()`
- `linspace()`
- `eye()`

```python
print(np.zeros(5))
print(np.zeros((2, 3)))

print(np.ones(4))
print(np.ones((2, 3)))

print(np.full(5, 7))
print(np.full((2, 3), 9))

print(np.arange(1, 10, 2))
print(np.linspace(0, 10, 5))

print(np.eye(3))
```

## 10. Random Arrays

- `np.random.rand()` → random decimal values from 0 to 1
- `np.random.randint()` → random integers
- `np.random.seed()` → repeatable random results

```python
np.random.seed(10)

print(np.random.rand(5))
print(np.random.rand(2, 3))
print(np.random.randint(1, 10, 5))
print(np.random.randint(1, 100, (3, 3)))
```

## 11. Reshape, Flatten and Ravel

```python
arr = np.arange(1, 7)

matrix = arr.reshape(2, 3)

print("Original:", arr)
print("Reshape:")
print(matrix)

print("Flatten:", matrix.flatten())
print("Ravel:", matrix.ravel())

print("Flatten with -1:", matrix.reshape(-1))
```

## 12. Element-wise Arithmetic

NumPy performs arithmetic element by element.

```python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** 2)

print(a + 5)
print(a * 2)
```

## 13. Comparisons and Conditions

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr > 25)
print(arr == 30)
print(arr != 30)
print(arr[arr > 25])
```

## 14. Aggregation Functions

Important:
- `sum`
- `mean`
- `max`
- `min`
- `median`
- `std`
- `var`

```python
arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Max:", np.max(arr))
print("Min:", np.min(arr))
print("Median:", np.median(arr))
print("Std:", np.std(arr))
print("Variance:", np.var(arr))
```

## 15. Axis

For a 2D array:
- `axis=0` → down the rows → column-wise result
- `axis=1` → across columns → row-wise result

```python
marks = np.array([
    [80, 70, 90],
    [60, 75, 85],
    [90, 85, 95]
])

print("Column sum:", np.sum(marks, axis=0))
print("Row sum:", np.sum(marks, axis=1))

print("Column mean:", np.mean(marks, axis=0))
print("Row mean:", np.mean(marks, axis=1))

print("Column max:", np.max(marks, axis=0))
print("Row min:", np.min(marks, axis=1))
```

## 16. Broadcasting

A scalar operation is applied to every element.

```python
marks = np.array([50, 60, 70, 80])

bonus = 5

print(marks + bonus)
print(marks - bonus)
print(marks * 2)
```

## 17. Concatenation

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate((a, b)))
```

## 18. Concatenation with 2D Arrays

```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

print("axis=0:")
print(np.concatenate((a, b), axis=0))

print("axis=1:")
print(np.concatenate((a, b), axis=1))
```

## 19. vstack and hstack

```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

print("vstack:")
print(np.vstack((a, b)))

print("hstack:")
print(np.hstack((a, b)))
```

## 20. split and array_split

```python
arr = np.arange(1, 9)

print(np.split(arr, 2))
print(np.array_split(arr, 3))
```

## 21. Transpose

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr)
print(arr.T)
```

## 22. Mathematical Functions

- `sqrt`
- `power`
- `abs`
- `round`
- `floor`
- `ceil`

```python
arr = np.array([-2.7, -1.2, 2.3, 4.8])

print("sqrt:", np.sqrt(np.array([4, 9, 16, 25])))
print("power:", np.power(np.array([2, 3, 4]), 2))
print("abs:", np.abs(arr))
print("round:", np.round(arr))
print("round 2:", np.round(arr, 2))
print("floor:", np.floor(arr))
print("ceil:", np.ceil(arr))
```

## 23. Logarithmic Functions

```python
arr = np.array([1, 10, 100, 1000])

print("log:", np.log(arr))
print("log10:", np.log10(arr))
```

## 24. Trigonometric Functions

```python
angles = np.array([0, np.pi / 2, np.pi])

print("sin:", np.sin(angles))
print("cos:", np.cos(angles))
print("tan:", np.tan(angles))
```

## 25. Percentiles

```python
marks = np.array([40, 50, 60, 70, 80, 90, 100])

print("25th:", np.percentile(marks, 25))
print("50th:", np.percentile(marks, 50))
print("75th:", np.percentile(marks, 75))
```

## 26. Unique Values and Counts

```python
arr = np.array([10, 20, 20, 30, 30, 30, 40])

print("Unique:", np.unique(arr))

values, counts = np.unique(arr, return_counts=True)

print("Values:", values)
print("Counts:", counts)
```

## 27. where()

### Two common forms

`np.where(condition)` → positions where condition is true

`np.where(condition, true_value, false_value)` → choose values based on condition

```python
arr = np.array([10, 60, 30, 80, 40])

print(np.where(arr > 50))

result = np.where(arr >= 50, "Pass", "Fail")
print(result)
```

## 28. Sorting

```python
arr = np.array([50, 20, 80, 10, 40])

print("np.sort:", np.sort(arr))

arr2 = arr.copy()
arr2.sort()

print("arr.sort:", arr2)
print("Original:", arr)
```

## 29. argsort()

Returns the indexes that would sort the array.

```python
marks = np.array([85, 60, 95, 75])
names = np.array(["Tejas", "Amit", "Rahul", "Sneha"])

index = np.argsort(marks)

print("Indexes:", index)
print("Sorted names:", names[index])
print("Sorted marks:", marks[index])
```

## 30. argmax and argmin

```python
marks = np.array([85, 60, 95, 75])

print("Maximum value:", np.max(marks))
print("Maximum index:", np.argmax(marks))

print("Minimum value:", np.min(marks))
print("Minimum index:", np.argmin(marks))
```

## 31. nonzero(), any(), all()

```python
arr = np.array([0, 5, 0, 10, 20])

print("Non-zero positions:", np.nonzero(arr))

marks = np.array([60, 70, 80])

print("Any > 75:", np.any(marks > 75))
print("All > 50:", np.all(marks > 50))
print("All > 75:", np.all(marks > 75))
```

## 32. Covariance and Correlation

```python
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([20, 40, 60, 80, 100])

print("Covariance:")
print(np.cov(hours, marks))

print("Correlation:")
print(np.corrcoef(hours, marks))
```

## 33. Linear Algebra

Important functions:
- `np.dot()`
- `np.matmul()`
- `@`
- `np.linalg.det()`
- `np.linalg.inv()`
- `np.linalg.solve()`
- `np.linalg.eig()`
- `np.linalg.norm()`

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)

print("matmul:")
print(np.matmul(A, B))

print("dot:")
print(np.dot(A, B))

print("Transpose:")
print(A.T)

print("Determinant:")
print(np.linalg.det(A))

print("Inverse:")
print(np.linalg.inv(A))

print("Norm:")
print(np.linalg.norm(np.array([3, 4])))
```

## 34. Solving Linear Equations

Example:

`2x + y = 5`

`x + y = 3`

```python
A = np.array([
    [2, 1],
    [1, 1]
])

B = np.array([5, 3])

solution = np.linalg.solve(A, B)

print("x =", solution[0])
print("y =", solution[1])
```

## 35. Eigenvalues and Eigenvectors

```python
A = np.array([
    [2, 0],
    [0, 3]
])

values, vectors = np.linalg.eig(A)

print("Eigenvalues:")
print(values)

print("Eigenvectors:")
print(vectors)
```
