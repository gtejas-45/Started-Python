```python
import matplotlib.pyplot as plt
```

```python
X =[1,2,3,4,5]
Y = [4, 2,1,3,5]
```

```python
plt.scatter(X,Y)
```

```python
plt.scatter(X,Y,marker = "d")
```

```python
plt.scatter(X,Y,marker = "o", color = 'red')#like that we have various marker read it on google
```

```python
plt.plot(X,Y)
```

```python
plt.plot(X,Y,marker = 'o')
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.title('line_graph')
```

```python
plt.plot(X,Y,marker = 'd',markerfacecolor = 'red')
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.title('line_graph')
```

```python
plt.plot(X,Y,marker = 'd',markerfacecolor = 'red',color = 'silver')
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.title('line_graph')
```

```python
plt.plot(X,Y,marker = 'd',markerfacecolor = 'red',color = 'silver',ls = ':',lw = 3)#we can write linestyle = ls and lw = linewidth
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.title('line_graph')
```

```python
plt.plot(X,Y,marker = 'd',markerfacecolor = 'red',color = 'silver',ls = '-',lw = 3,markersize = 5)#we can write linestyle = ls and lw = linewidth
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.title('line_graph')
```

```python
plt.plot(X,Y,marker = 'd',markerfacecolor = 'red',color = 'silver',ls = '-',lw = 3,markersize = 5)#we can write linestyle = ls and lw = linewidth
plt.xlim([0,6])
plt.ylim([0,6])
plt.title('line_graph')#initially it saves the space so iyt start from the 0.5 rather than the 0 so for removing this we use the xlim funtion 
```

```python
import numpy as np
```

```python
x = np.arange(0,np.pi * 2,0.1)
y = np.sin(x)
plt.plot(x,y,ls=":")
# plt.xticks([1,2,3,4,5],['a','b','c','d','e'])
```

```python
y = np.cos(x)
plt.plot(x,y,ls=":")
```

```python
y = np.tan(x)
plt.plot(x,y,ls=":")
```
